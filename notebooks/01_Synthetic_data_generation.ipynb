```python
# =============================================
# Cell 1: Import Libraries
# =============================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime

print("🚀 Transportation Security System")
print("📅 Started:", datetime.now().strftime("%Y-%m-%d %H:%M:%S"))
print("✅ All imports successful!\n")

# =============================================
# Cell 2: Generate GPS Data
# =============================================
def generate_gps_data(n=200):
    """Generate synthetic GPS and sensor data"""
    np.random.seed(42)
    
    data = {
        'timestamp': pd.date_range('2024-01-01', periods=n, freq='1s'),
        'speed': np.random.normal(60, 15, n),
        'latitude': np.random.normal(40.7128, 0.005, n),
        'longitude': np.random.normal(-74.0060, 0.005, n),
        'acceleration': np.random.normal(0, 2, n),
        'braking': np.abs(np.random.normal(0, 1, n))
    }
    return pd.DataFrame(data)

# =============================================
# Cell 3: Create and View Data
# =============================================
df = generate_gps_data(200)
print("📊 Data Shape:", df.shape)
df.head()

# =============================================
# Cell 4: Data Statistics
# =============================================
print("\n📈 Summary Statistics:")
df.describe()

# =============================================
# Cell 5: Plot Speed Data
# =============================================
plt.figure(figsize=(12, 4))
plt.plot(df['timestamp'], df['speed'], linewidth=2)
plt.title('GPS Speed Data', fontsize=14)
plt.xlabel('Time')
plt.ylabel('Speed (km/h)')
plt.grid(True, alpha=0.3)
plt.show()
```
